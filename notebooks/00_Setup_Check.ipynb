```python
# 4. Reranker
from sentence_transformers import CrossEncoder

reranker = CrossEncoder("BAAI/bge-reranker-v2-m3", device="cuda")
scores = reranker.predict([
    ("What is Knowledge OS?", "Knowledge OS is a local enterprise RAG platform.")
])
print(scores)
```
